```python
  import pandas as pd
  url = "https://raw.githubusercontent.com/pandas-dev/pandas/main/pandas/tests/io/data/csv/tips.csv"
  tips = pd.read_csv(url)
  tips.head(); tips.info(); tips.describe()
```

```python
  # WHERE + nový stĺpec
  tips["tip_pct"] = tips["tip"] / tips["total_bill"]
  tips[(tips["time"] == "Dinner") & (tips["size"] >= 4)]
```

```python
  # GROUP BY
  tips.groupby("day").agg(n=("tip", "size"), avg_tip_pct=("tip_pct", "mean")).sort_values("avg_tip_pct")
  tips.groupby(["day", "smoker"])["tip_pct"].mean().unstack()
```

```python
  # JOIN
  days = pd.DataFrame({"day": ["Thur", "Fri", "Sat", "Sun"], "weekend": [False, False, True, True]})
  merged = tips.merge(days, on="day", how="left")
  merged.groupby("weekend")["tip_pct"].mean()
```

Weekend tip percentage (16.0 %) is practically the same as on weekdays (16.3 %); the difference is not significant (Welch t-test, p ≈ 0.60)
